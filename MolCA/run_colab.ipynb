## Training

```python
%env TOKENIZERS_PARALLELISM=false

!python3.8 MolCA/custom_train.py \
    --checkpoint checkpoints/chebi.ckpt \
    --peft_dir checkpoints/chebi_lora \
    --gnn_ckpt checkpoints/graphcl_80.pth \
    --train_file data/smiles/train_smiles.tsv \
    --val_file data/smiles/validation_smiles.tsv \
    --output_dir checkpoints/finetuned \
    --batch_size 10 \
    --eval_batch_size 10 \
    --grad_accum_steps 3 \
    --max_epochs 50 \
    --warmup_epochs 2 \
    --patience 2 \
    --eval_every 5 \
    --text_max_len 384 \
    --num_beams 5 \
    --seed 42 \
    --device cuda \
    --val_percentage 100 \
    --min_lr 1e-6 \
    --lr_lora 1e-5 \
    --lr_proj 1e-4 \
    --wd_lora 0.0 \
    --wd_proj 0.01 \
    --t0_epochs 5 \
    --t_mult 2 \
    --quick_val_fraction 0.10 \
    --quick_eval_every 0.0
```
